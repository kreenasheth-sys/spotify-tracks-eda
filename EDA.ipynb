```python
# Lets undertsnad the plot, The middle line is Q2 values or median value. 
# First line of the box is Q1 value and last line of the box is Q3 value. And the width of this box is inter-quartile range. 
# The line at the end are called whiskers and these are the min and max values determined by the interquartile range. 
# The left line is derived by subtracting 1.5 from the Q1 and right one or max is derived by adding 1.5 to q3. 
# Anything outside this are called outliers and we can see that at the start this has outliers
```

```python
plt.figure(figsize=(10, 4))

df['energy'].hist(bins=30)

plt.xlabel('Energy')
plt.ylabel('Frequency')
plt.title('Distribution of Energy')

plt.show()
```

```python
# Finding:
# The distribution of energy is concentrated towards higher values,
# indicating that most tracks in the dataset have relatively high energy.
# The distribution is negatively skewed, with fewer tracks having low energy values.
```

```python
plt.figure(figsize=(10, 4))

df['popularity'].hist(bins=30)

plt.xlabel('Popularity')
plt.ylabel('Frequency')
plt.title('Distribution of Track Popularity')

plt.show()
```

```python
# Finding:
# A large number of tracks have popularity values close to 0.
# Most of the remaining tracks have popularity scores between 20 and 70,
# while very highly popular tracks are relatively less common.
```

```python
plt.figure(figsize=(10, 3))

sns.boxplot(x=df['popularity'])

plt.title('Boxplot of Track Popularity')
plt.xlabel('Popularity')

plt.show()
```

```python
# Finding:
# The popularity values show a wide spread across the dataset.
# The middle 50% of tracks have popularity values roughly between 17 and 50.
# The median popularity is around 35.
# A high-value outlier is also observed.
```

```python
# Multivariate Analysis
```

```python
df.columns
```

```python
df_corr = df[['popularity',
       'duration_ms',  'danceability', 'energy', 'key', 'loudness',
       'mode', 'speechiness', 'acousticness', 'instrumentalness', 'liveness',
       'valence', 'tempo', 'time_signature']].corr()
```

```python
print(df_corr)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    df_corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
# Findings:
# 1. Energy and loudness show a strong positive correlation (0.76),
#    indicating that more energetic tracks tend to be louder.
#
# 2. Energy and acousticness have a strong negative correlation (-0.73),
#    suggesting that energetic tracks tend to be less acoustic.
#
# 3. Loudness and acousticness show a moderately strong negative
#    correlation (-0.59).
#
# 4. Danceability and valence have a moderate positive correlation (0.48),
#    suggesting that more danceable tracks tend to have higher valence.
#
# 5. Popularity has weak correlations with most audio features,
#    indicating that no single feature has a strong linear relationship
#    with popularity.
#
# 6. Most remaining variables have weak correlations with each other.
```
